```python
import random
graph = {
    'A': {'B': 2, 'C': 5},
    'B': {'A': 2, 'D': 3},
    'C': {'A': 5, 'D': 1},
    'D': {'B': 3, 'C': 1}
}
num_ants = 10
num_iterations = 50
alpha = 1.0
beta = 2.0
evaporation = 0.5
Q = 100
pheromone = {}
for node in graph:
    for neighbor in graph[node]:
        pheromone[(node, neighbor)] = 1.0
def choose_next(current, visited):
    candidates = [
        node for node in graph[current]
        if node not in visited
    ]
    if not candidates:
        return None
    probabilities = []
    for node in candidates:
        distance = graph[current][node]
        heuristic = 1 / distance
        value = (pheromone[(current, node)] ** alpha) * \
                (heuristic ** beta)
        probabilities.append(value)
    total = sum(probabilities)
    probabilities = [p / total for p in probabilities]
    return random.choices(candidates, weights=probabilities, k=1)[0]
def construct_path(start, end):
    current = start
    path = [current]
    visited = {current}
    distance = 0
    while current != end:
        next_node = choose_next(current, visited)
        if next_node is None:
            return None, float('inf')
        distance += graph[current][next_node]
        path.append(next_node)
        visited.add(next_node)
        current = next_node
    return path, distance
def aco_shortest_path(start, end):
    best_path = None
    best_distance = float('inf')
    for iteration in range(num_iterations):
        all_paths = []
        for ant in range(num_ants):
            path, distance = construct_path(start, end)
            if path is not None:
                all_paths.append((path, distance))
                if distance < best_distance:
                    best_distance = distance
                    best_path = path
        for edge in pheromone:
            pheromone[edge] *= (1 - evaporation)
        for path, distance in all_paths:
            deposit = Q / distance
            for i in range(len(path) - 1):
                edge = (path[i], path[i + 1])
                pheromone[edge] += deposit
                reverse_edge = (path[i + 1], path[i])
                if reverse_edge in pheromone:
                    pheromone[reverse_edge] += deposit
    return best_path, best_distance
start = 'A'
end = 'D'

path, distance = aco_shortest_path(start, end)

print("Shortest Path:", " -> ".join(path))
print("Shortest Distance:", distance)
```
